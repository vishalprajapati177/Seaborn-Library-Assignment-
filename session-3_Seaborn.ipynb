```python
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#Question-1
# Swiggy-style delivery time data
delivery_times = [
    25, 30, 35, 28, 32,
    40, 45, 30, 27, 35,
    38, 42, 29, 31, 36,
    33, 25, 40, 45, 37,
    28, 34, 39, 41, 30,
    32, 35, 27, 43, 38
]

# Create histogram
sns.histplot(
    delivery_times,
    bins=8
)

# Add labels and title
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")
plt.title("Swiggy Delivery Time Distribution")

# Display chart
plt.show()
```

```python
#Question-2
# Daily step counts for one month
steps = [
    5000, 6500, 7000, 5500, 8000,
    7500, 6200, 9000, 8500, 7200,
    6800, 5600, 7400, 8100, 9200,
    6000, 6700, 7300, 7800, 8800,
    5400, 5900, 7600, 8300, 9500,
    7000, 6500, 8700, 9100, 7200
]

# Create KDE plot
sns.kdeplot(
    steps,
    fill=True
)

plt.xlabel("Daily Steps")
plt.ylabel("Density")
plt.title("Daily Step Count Distribution")

plt.show()
```

```python
#Question-2
# Monthly Zomato spending
monthly_spending = [
    250, 450, 700, 1200, 850,
    500, 1500, 950, 650, 1800,
    1100, 750, 400, 1300, 900,
    600, 2000, 1050, 550, 1400,
    800, 1700, 1250, 700, 1000,
    1600, 450, 1350, 900, 1150
]

# Create one figure and one Axes
fig, ax = plt.subplots(figsize=(10, 5))

# Histogram
sns.histplot(
    monthly_spending,
    bins=10,
    ax=ax
)

# KDE curve on the same Axes
sns.kdeplot(
    monthly_spending,
    ax=ax
)

# Labels and title
ax.set_xlabel("Monthly Spending (₹)")
ax.set_ylabel("Density / Frequency")
ax.set_title("Zomato Monthly Spending Distribution")

plt.show()
```

```python
#Question-4

# Sample IPL match scores
scores = [
    120, 145, 160, 175, 190,
    135, 155, 180, 200, 165,
    142, 158, 172, 188, 210,
    125, 150, 168, 182, 195,
    130, 148, 162, 178, 185,
    205, 220, 140, 155, 170
]


# -----------------------------
# 5 Bins
# -----------------------------

plt.figure()

sns.histplot(
    scores,
    bins=5
)

plt.xlabel("IPL Score")
plt.ylabel("Number of Matches")
plt.title("IPL Score Distribution - 5 Bins")

plt.show()


# -----------------------------
# 10 Bins
# -----------------------------

plt.figure()

sns.histplot(
    scores,
    bins=10
)

plt.xlabel("IPL Score")
plt.ylabel("Number of Matches")
plt.title("IPL Score Distribution - 10 Bins")

plt.show()


# -----------------------------
# 20 Bins
# -----------------------------

plt.figure()

sns.histplot(
    scores,
    bins=20
)

plt.xlabel("IPL Score")
plt.ylabel("Number of Matches")
plt.title("IPL Score Distribution - 20 Bins")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv(r"C:\Users\VISHAL PRAJAPATI\Downloads\instagram_users.csv")

# Display first 5 rows
print(df.head())

# Create scatterplot
sns.scatterplot(
    x="followers",
    y="posts",
    data=df,
    color="red"
)

# Add labels and title
plt.xlabel("Followers")
plt.ylabel("Number of Posts")
plt.title("Instagram Followers vs Number of Posts")

# Display plot
plt.show()
```

```python
#Question-2
# Load CSV
df = pd.read_csv(R"C:\Users\VISHAL PRAJAPATI\Downloads\zomato_daily_orders.csv")

# Display data
print(df.head())

# Create line plot
sns.lineplot(
    x="day",
    y="orders",
    data=df,
    color="blue"
)

# Add labels and title
plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.title("Daily Zomato Orders")

# Display plot
plt.show()
```

```python
#Question-3
# Load CSV
df = pd.read_csv(R"C:\Users\VISHAL PRAJAPATI\Downloads\movies_genres.csv")

# Display first 5 rows
print(df.head())

# Create relational plot
g = sns.relplot(
    x="rating",
    y="reviews",
    hue="genre",
    data=df
)

# Add title
g.fig.suptitle(
    "Movie Ratings vs Number of Reviews",
    y=1.02
)

# Display plot
plt.show()
```

```python
#Question-4
# Load CSV
df = pd.read_csv(R"C:\Users\VISHAL PRAJAPATI\Downloads\youtube_exam_scores.csv")

# Display first 5 rows
print(df.head())

# Create regression plot
sns.regplot(
    x="youtube_hours",
    y="exam_score",
    data=df,
    color="grey"
)

# Add labels and title
plt.xlabel("Hours Spent on YouTube")
plt.ylabel("Exam Score")
plt.title("YouTube Usage vs Exam Scores")

# Display plot
plt.show()
```

```python
# Question-1
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Flipkart CSV
df = pd.read_csv(R"C:\Users\VISHAL PRAJAPATI\Downloads\flipkart_ratings_prices.csv")

# Display first 5 rows
print(df.head())

# Create scatter jointplot
sns.jointplot(
    x="rating",
    y="price",
    data=df,
    kind="scatter",
    color="orange"
)

# Display plot

plt.show()
```

```python
# Question-4

# Sample Zomato restaurant review data
data = {
    "rating": [
        3.1, 3.5, 4.0, 4.2, 4.5,
        3.8, 4.1, 4.3, 3.7, 4.6,
        3.2, 3.9, 4.4, 4.7, 3.6,
        4.0, 4.2, 4.5, 3.4, 4.8
    ],

    "reviews": [
        120, 250, 500, 800, 1200,
        400, 700, 950, 350, 1500,
        180, 600, 1100, 1800, 300,
        750, 1000, 1300, 220, 2000
    ]
}

# Create DataFrame
df = pd.DataFrame(data)


# -------------------------------
# 1. Scatter Jointplot
# -------------------------------

sns.jointplot(
    x="rating",
    y="reviews",
    data=df,
    kind="scatter",
    color="pink"
)

plt.show()


# -------------------------------
# 2. Hex Jointplot
# -------------------------------

sns.jointplot(
    x="rating",
    y="reviews",
    data=df,
    kind="hex",
    color="red"
)

plt.show()


# -------------------------------
# 3. KDE Jointplot
# -------------------------------

sns.jointplot(
    x="rating",
    y="reviews",
    data=df,
    kind="kde",
    color="yellow"
)

plt.show()
```

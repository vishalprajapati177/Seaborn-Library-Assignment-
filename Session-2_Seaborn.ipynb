```python
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Sample Swiggy-style data
data = {
    "payment_method": [
        "UPI", "Card", "UPI", "Cash",
        "UPI", "Card", "UPI", "Cash",
        "Card", "UPI"
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create countplot
sns.countplot(
    x="payment_method",
    data=df,color="orange"
)

# Labels and title
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.title("Orders by Payment Method")

# Display chart
plt.show()
```

```python
#Question-2
# Zomato-style sample data
data = {
    "category": [
        "Pizza", "Pizza", "Pizza",
        "Burger", "Burger", "Burger",
        "Biryani", "Biryani", "Biryani",
        "South Indian", "South Indian", "South Indian"
    ],

    "delivery_time": [
        30, 35, 40,
        20, 25, 30,
        40, 45, 50,
        25, 30, 35
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Average delivery time
sns.barplot(
    x="category",
    y="delivery_time",
    data=df,
    estimator="mean",
    color="red"
)

plt.xlabel("Food Category")
plt.ylabel("Average Delivery Time (Minutes)")
plt.title("Average Delivery Time by Food Category")

plt.show()
```

```python
#Question--3
# Sample order data
data = {
    "category": [
        "Pizza", "Pizza", "Pizza",
        "Burger", "Burger", "Burger",
        "Biryani", "Biryani", "Biryani",
        "South Indian", "South Indian", "South Indian"
    ],

    "orders": [
        100, 150, 200,
        120, 180, 160,
        200, 220, 180,
        90, 110, 130
    ]
}

df = pd.DataFrame(data)

# Total orders by category
sns.barplot(
    x="category",
    y="orders",
    data=df,
    estimator="sum"
)

plt.xlabel("Food Category")
plt.ylabel("Total Number of Orders")
plt.title("Total Orders by Food Category")

plt.show()
```

```python
# Question-4
# BookMyShow-style movie data
data = {
    "genre": [
        "Action",
        "Comedy",
        "Drama",
        "Action",
        "Comedy",
        "Action",
        "Romance",
        "Drama",
        "Action",
        "Comedy"
    ]
}

df = pd.DataFrame(data)

# Count movies by genre
sns.countplot(
    x="genre",
    data=df,
    color="green"
)

plt.xlabel("Movie Genre")
plt.ylabel("Number of Movies")
plt.title("Movie Genre Distribution")

plt.show()
```

```python
# Question-1
import pandas as pd
import numpy as np

# Make random data reproducible
np.random.seed(42)

# Create random data
delivery_time = np.random.randint(20, 61, 20)

order_value = np.random.randint(200, 1501, 20)

tip_amount = np.random.randint(10, 101, 20)

rating = np.round(
    np.random.uniform(3.0, 5.0, 20),
    1
)

delivery_distance = np.round(
    np.random.uniform(1, 15, 20),
    1
)

# Create DataFrame
df = pd.DataFrame({
    "delivery_time": delivery_time,
    "order_value": order_value,
    "tip_amount": tip_amount,
    "rating": rating,
    "delivery_distance": delivery_distance
})

# Display first 5 rows
print("First 5 rows:")
print(df.head())

# Calculate correlation matrix
correlation_matrix = df.corr()

# Display correlation matrix
print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
# Question-2
import seaborn as sns
import matplotlib.pyplot as plt

# Reproducible random data
np.random.seed(42)

# Create data
df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 61, 20),
    "order_value": np.random.randint(200, 1501, 20),
    "tip_amount": np.random.randint(10, 101, 20),
    "rating": np.round(np.random.uniform(3.0, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 1)
})

# Calculate correlation
correlation_matrix = df.corr()

# Create heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
)

# Add title
plt.title("Food Delivery Correlation Matrix")

# Display
plt.show()
```

```python
# Question-3
# Reproducible random data
np.random.seed(42)

# Create DataFrame
df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 61, 20),
    "order_value": np.random.randint(200, 1501, 20),
    "tip_amount": np.random.randint(10, 101, 20),
    "rating": np.round(np.random.uniform(3.0, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 1)
})

# Calculate correlation matrix
correlation_matrix = df.corr()

# Heatmap with diverging palette
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Food Delivery Correlation Heatmap")

plt.show()

# Find strongest negative correlation
corr_pairs = correlation_matrix.unstack()

negative_pairs = corr_pairs[
    corr_pairs < 0
]

print("\nStrongest negative correlation:")
print(negative_pairs.sort_values().head(1))
```

```python
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#Question-1
# Load CSV file

df = pd.read_csv(r"C:\Users\VISHAL PRAJAPATI\Downloads\flipkart_products.csv")

# Display first 5 rows
print(df.head())

# Create boxplot
sns.boxplot(
    y=df["price"]
)

# Add label and title
plt.ylabel("Price (₹)")
plt.title("Flipkart Product Price Distribution")

# Display chart
plt.show()
```

```python
#Question-2
# Sample Zomato-style data
data = {
    "category": [
        "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Fast Food",
        "Desserts", "Desserts", "Desserts",
        "Desserts", "Desserts"
    ],

    "order_amount": [
        250, 300, 350, 400, 280,
        150, 200, 220, 180, 250
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create violin plot
sns.violinplot(
    x="category",
    y="order_amount",
    data=df
)

# Labels and title
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.title("Order Amount Distribution by Restaurant Category")

plt.show()
```

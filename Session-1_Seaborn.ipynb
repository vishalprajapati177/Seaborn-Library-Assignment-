```python
#Question-1
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

print("Seaborn version:", sns.__version__)
print("Matplotlib version:", matplotlib.__version__)
```

```python
#Question-2
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(
    x=days,
    y=temps,
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")
plt.grid()

plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temps = [32, 34, 33, 35, 36, 37, 34]

sns.relplot(
    x=days,
    y=temps,
    kind="line",
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
# Question-4
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temps = [32, 34, 33, 35, 36, 37, 34]


# --------------------------------
# Theme 1: Darkgrid
# --------------------------------

sns.set_theme(style="darkgrid")

plt.figure()

sns.lineplot(
    x=days,
    y=temps,
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Darkgrid")

plt.savefig(
    "temperature_darkgrid.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# --------------------------------
# Theme 2: Whitegrid
# --------------------------------

sns.set_theme(style="whitegrid")

plt.figure()

sns.lineplot(
    x=days,
    y=temps,
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Whitegrid")

plt.savefig(
    "temperature_whitegrid.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# --------------------------------
# Theme 3: Ticks
# --------------------------------

sns.set_theme(style="ticks")

plt.figure()

sns.lineplot(
    x=days,
    y=temps,
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Ticks")

plt.savefig(
    "temperature_ticks.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

```python
# Question-1
import seaborn as sns
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
streams = [120, 150, 135, 180, 210, 260, 230]

sns.set(context="poster")

sns.barplot(x=days, y=streams,color="orange")

plt.title("Spotify Songs Streamed Per Day")
plt.xlabel("Day")
plt.ylabel("Songs Streamed")

plt.show()
```

```python
# Question-2

cuisine = [
    "Indian", "Chinese", "Indian", "Italian",
    "Chinese", "Indian", "Italian", "Mexican",
    "Indian", "Chinese", "Italian", "Indian"
]

sns.set(context="notebook")

sns.countplot(x=cuisine, palette="colorblind")

plt.title("Food Orders by Cuisine Type")
plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")

plt.show()
```

```python
# Question-3

days = list(range(1, 15))

steps = [
    5200, 6100, 5800, 7200, 6800, 7500, 8100,
    6300, 6900, 7700, 8200, 8600, 7900, 9100
]

sns.set(context="notebook", palette="muted")

sns.lineplot(x=days, y=steps, marker="o")

plt.title("Daily Step Count - 2 Weeks")
plt.xlabel("Day")
plt.ylabel("Steps")

plt.show()
```

```python
# Question-4

categories = ["A", "B", "C", "D"]
sales = [250, 320, 280, 400]

sns.set(context="poster", palette="colorblind")

sns.barplot(x=categories, y=sales,color="yellow")

plt.title("Category Sales")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
#Question-5

categories = [
    "Mobiles",
    "Electronics",
    "Fashion",
    "Home",
    "Beauty"
]

sales = [850, 620, 540, 430, 350]

sns.set(
    context="talk",
    style="whitegrid",
    palette="deep",
    font_scale=1.1
)

sns.barplot(x=categories, y=sales)

plt.title("Flipkart Sales Dashboard")
plt.xlabel("Product Category")
plt.ylabel("Sales (₹ Lakhs)")

plt.show()
```

```python
# Question-1
import seaborn as sns
import matplotlib.pyplot as plt

categories = ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"]
orders = [250, 320, 280, 190, 150]

sns.set_theme(style="whitegrid")

sns.barplot(x=categories, y=orders,color="grey")

plt.title("Food Category Orders")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")

plt.savefig("food_orders_300dpi.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
#Question-2

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5200, 6100, 5800, 7200, 6800, 7500, 8200]

sns.set_theme(style="whitegrid")

sns.lineplot(x=days, y=steps, marker="o")

plt.title("Daily Step Count - One Week")
plt.xlabel("Day")
plt.ylabel("Steps")

plt.savefig(
    "daily_steps.svg",
    format="svg",
    bbox_inches="tight"
)

plt.show()
```

```python
#Question-3

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [450, 520, 480, 610, 680, 750]

sns.set_theme(
    style="darkgrid",
    palette="muted"
)

sns.barplot(x=months, y=sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig(
    "monthly_sales_1200dpi.png",
    dpi=1200,
    bbox_inches="tight"
)

plt.show()
```

```python
#Question-4
def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        palette="deep",
        font_scale=1.2
    )


apps = ["Instagram", "YouTube", "WhatsApp", "Chrome", "Spotify"]
usage = [30, 25, 20, 15, 10]

apply_presentation_theme()

plt.figure(figsize=(8, 6))

plt.pie(
    usage,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

plt.savefig(
    "top_5_apps.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Question-5
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd



data = pd.DataFrame({
    "Price": [100, 200, 300, 400, 500],
    "Rating": [4.1, 4.3, 4.5, 4.2, 4.6],
    "Reviews": [120, 250, 350, 180, 420],
    "Sales": [80, 150, 220, 130, 280]
})

correlation = data.corr()

sns.set_theme(style="darkgrid")

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Product Data Correlation Heatmap")

plt.savefig(
    "product_heatmap.svg",
    format="svg",
    dpi=1200,
    bbox_inches="tight"
)

plt.show()
```

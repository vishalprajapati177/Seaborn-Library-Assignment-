```python
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#Question-1


# Load the built-in tips dataset
tips = sns.load_dataset('tips')

# Display first 5 rows
print(tips.head())

# Create basic pairplot
sns.pairplot(tips)

# Display the plot
plt.show()
```

```python
#Question-2

# Load tips dataset
tips = sns.load_dataset("tips")

# Create pairplot with sex-based colors
sns.pairplot(
    tips,
    hue="sex"
)

# Display plot
plt.show()
```

```python
#Question-3

# Load tips dataset
tips = sns.load_dataset("tips")

# Pairplot with smoker as hue
# and KDE on diagonal
sns.pairplot(
    tips,
    hue="smoker",
    diag_kind="kde"
)

# Display plot
plt.show()
```

```python
#Question-4

# Load tips dataset
tips = sns.load_dataset("tips")

# Create pairplot using only two numerical columns
# and color points according to meal time
sns.pairplot(
    tips,
    vars=["total_bill", "tip"],
    hue="time"
)

# Display plot
plt.show()
```
